# IPL Analytics Engine EDA and Feature Engineering

This notebook is the beginner-friendly starting point for the project. It shows how to load IPL data, inspect it, and turn raw match rows into useful cricket features.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import plotly.express as px

from utils.preprocessing import build_demo_data, load_ipl_data
from utils.feature_engineering import engineer_ball_by_ball_features, build_training_frame

pd.set_option('display.max_columns', 200)
```

```python
data_dir = Path('..') / 'data'
matches_path = data_dir / 'matches.csv'
deliveries_path = data_dir / 'deliveries.csv'

if matches_path.exists() and deliveries_path.exists():
    matches, deliveries = load_ipl_data(matches_path, deliveries_path)
else:
    matches, deliveries = build_demo_data()

matches.head(), deliveries.head()
```

## 2. Basic Data Checks

This helps you understand what columns exist and how much data you have.

```python
print('Matches shape:', matches.shape)
print('Deliveries shape:', deliveries.shape)
print('Match columns:', list(matches.columns))
print('Delivery columns:', list(deliveries.columns))
```

## 3. Feature Engineering

Now we convert the raw rows into prediction-friendly numbers like run rate, wickets remaining, and pressure index.

```python
engineered = engineer_ball_by_ball_features(deliveries, matches)
training_frame = build_training_frame(matches, deliveries)

sample_match_id = engineered['match_id'].iloc[0]
sample = engineered[engineered['match_id'] == sample_match_id].copy()
sample['ball_number'] = range(1, len(sample) + 1)
sample['innings_runs'] = sample['total_runs'].cumsum()

engineered.head(), training_frame.head(), px.line(sample, x='ball_number', y='innings_runs', title='Run Progression')
```
